Questão 1: No bloco 3, a função train_test_split é executada com o parâmetro test_size=0.2. Explique qual é a finalidade de separar os dados dessa forma e por que o conjunto de teste não deve ser utilizado na etapa de treinamento.

**Resposta:** O parâmetro test_size=0.2 separa 20% dos dados exclusivamente para avaliação. Essa divisão é importante para verificar a capacidade de generalização do modelo em dados que ele nunca teve contato. Utilizar o conjunto de teste durante o treinamento pode comprometer a avaliação, pois o modelo pode memorizar os exemplos (overfitting) e apresentar um resultado artificialmente elevado, sem realmente conseguir generalizar para novos dados.

Questão 2: O parâmetro random_state=42 foi definido na divisão dos dados. Qual é a utilidade prática desse argumento em experimentos de Machine Learning?

**Resposta:** O random_state determina uma semente fixa para o gerador de números aleatórios usado na divisão dos dados. Dessa maneira, cada execução do código mantém exatamente a mesma separação entre os conjuntos de treino e teste, permitindo que o experimento seja reproduzido por diferentes pessoas e máquinas.

Questão 3: No bloco 4, inicializamos o classificador com kernel='linear'. Em que tipo de cenário ou característica dos dados o uso de um kernel linear é recomendado, e o que o algoritmo busca encontrar durante o método .fit()?

**Resposta:** O kernel linear é indicado quando as classes podem ser separadas linearmente, isto é, quando existe um hiperplano capaz de dividi-las. O método .fit() corresponde à fase de treinamento, na qual o algoritmo ajusta esse hiperplano utilizando os dados de treino e buscando maximizar a margem entre as classes.

Questão 4: O notebook utiliza o conjunto de dados Iris. Descreva brevemente o que representam as variáveis X (iris.data) e y (iris.target) dentro do escopo deste dataset.

**Resposta:** X (iris.data) armazena os atributos de entrada de cada amostra, como as medidas das sépalas e pétalas das flores. Já y (iris.target) guarda os rótulos de saída, que representam a espécie de cada amostra e correspondem ao resultado que o modelo deve aprender a prever.


Questão 5: No bloco final de código, a execução do comando print("Valores preditos:", y_pred) e print("Valores reais:", y_test) gerou os arrays correspondentes. Analise a saída obtida pelo modelo e comente o resultado da acurácia calculada (100.00%). Esse comportamento de acurácia perfeita é comum em cenários reais complexos? Justifique.

**Resposta:** Uma acurácia de 100% indica que o modelo classificou corretamente todas as amostras presentes no conjunto de teste. Em datasets simples, como o Iris, esse resultado pode acontecer de forma legítima devido à boa separação entre as classes. Porém, em situações reais e mais complexas, esse desempenho é menos comum e deve ser analisado, pois pode indicar overfitting, quando o modelo se adapta excessivamente aos dados de treino e apresenta menor capacidade de generalização.
