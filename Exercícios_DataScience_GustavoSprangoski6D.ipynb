3. Encontre as ra´ızes da fun¸c˜ao f(x) = x
4 − 3x
3 + 2 utilizando SciPy.

```python
from scipy.optimize import fsolve
import numpy as np

# define a funcao
def funcao(x):
    return x**4 - 3*x**3 + 2

# testa varios pontos de partida pra achar raizes diferentes
raizes = set()
for inicio in np.linspace(-2, 4, 30):
    res = fsolve(funcao, inicio, full_output=True)
    if res[2] == 1:
        raizes.add(round(res[0][0], 5))

print(sorted(raizes))
```

19. Crie um DataFrame a partir de um dicion´ario de listas e calcule a m´edia
de cada coluna num´erica.

```python
import pandas as pd

dados = {
    'Nome':        ['Ana', 'Bruno', 'Carlos', 'Diana'],
    'Idade':       [23, 35, 45, 28],
    'Salario':     [3500, 5000, 7000, 4200],
    'Experiencia': [2, 8, 15, 5]
}

tabela = pd.DataFrame(dados)

# seleciona so as colunas numericas e calcula a media de cada uma
medias = tabela.select_dtypes(include='number').mean()
print(medias)
```

20. Agrupe os dados por uma coluna categ´orica e calcule a mediana de uma
coluna num´erica.


```python
import pandas as pd

dados = {
    'Produto':   ['Arroz', 'Feijao', 'Leite', 'Queijo', 'Oleo', 'Manteiga'],
    'Categoria': ['Graos', 'Graos', 'Laticinios', 'Laticinios', 'Outros', 'Laticinios'],
    'Preco':     [5.50, 8.90, 4.20, 18.00, 7.30, 12.00]
}

tabela = pd.DataFrame(dados)

# agrupa pela categoria e calcula a mediana do preco em cada grupo
mediana_categoria = tabela.groupby('Categoria')['Preco'].median()
print(mediana_categoria)
```

18. Crie um gr´afico de barras com valores positivos e negativos e use diferentes
