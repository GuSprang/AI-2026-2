21.Busca Binária
o Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python
# lista ordenada para buscar
lista = [1, 3, 5, 7, 9, 11, 13]

# pede o numero a buscar
alvo = int(input("digite o numero a buscar: "))

# define o inicio e o fim da busca
inicio = 0
fim = len(lista) - 1
encontrado = False

# fica buscando enquanto houver espaco
while inicio <= fim:
    # pega o elemento do meio
    meio = (inicio + fim) // 2
    if lista[meio] == alvo:
        print("encontrado na posicao:", meio)
        encontrado = True
        break
    elif lista[meio] < alvo:
        inicio = meio + 1
    else:
        fim = meio - 1

# avisa se nao encontrou
if not encontrado:
    print("nao encontrado")
```

divide a lista ao meio toda hora e decide se busca na metade da esquerda ou da direita ate achar ou esgotar

22. Criptografia de Cifra de César
o Crie uma função para criptografar uma mensagem usando a cifra de
César

```python
# funcao que cifra a mensagem
def cifra_cesar(mensagem, chave):
    resultado = ""
    for letra in mensagem:
        # so muda letras, nao pontuacao
        if letra.isalpha():
            base = ord('a') if letra.islower() else ord('A')
            nova = chr((ord(letra) - base + chave) % 26 + base)
            resultado = resultado + nova
        else:
            resultado = resultado + letra
    return resultado

# pede a mensagem e a chave
mensagem = input("digite a mensagem: ")
chave = int(input("digite a chave: "))

# mostra a mensagem cifrada
print("mensagem cifrada:", cifra_cesar(mensagem, chave))
```

percorre cada letra e desloca ela no alfabeto pela chave pra criptografar

24.Validação de CPF
o Implemente uma função para validar um CPF brasileiro.

```python
# funcao que valida o cpf
def validar_cpf(cpf):
    # remove pontos e traco
    cpf = cpf.replace(".", "").replace("-", "")

    # cpf precisa ter 11 digitos
    if len(cpf) != 11:
        return False

    # calcula o primeiro digito verificador
    soma = 0
    for i in range(9):
        soma = soma + int(cpf[i]) * (10 - i)
    resto = soma % 11
    if resto < 2:
        digito1 = 0
    else:
        digito1 = 11 - resto
    if digito1 != int(cpf[9]):
        return False

    # calcula o segundo digito verificador
    soma = 0
    for i in range(10):
        soma = soma + int(cpf[i]) * (11 - i)
    resto = soma % 11
    if resto < 2:
        digito2 = 0
    else:
        digito2 = 11 - resto

    return digito2 == int(cpf[10])

# pede o cpf
cpf = input("digite o cpf: ")

# mostra se e valido ou nao
if validar_cpf(cpf):
    print("cpf valido")
else:
    print("cpf invalido")
```

calcula os dois digitos verificadores com somas ponderadas e compara com os digitos reais do cpf digitado

31.Ordenação por Bolha
o Implemente o algoritmo de ordenação por bolha (Bubble Sort).


```python
# lista para ordenar
numeros = [5, 3, 8, 1, 9, 2]
n = len(numeros)

# percorre a lista varias vezes
for i in range(n):
    for j in range(0, n - i - 1):
        # troca se o da esquerda for maior
        if numeros[j] > numeros[j + 1]:
            temp = numeros[j]
            numeros[j] = numeros[j + 1]
            numeros[j + 1] = temp

# mostra a lista ordenada
print("lista ordenada:", numeros)
```

compara pares vizinhos e troca se estiverem na ordem errada, repete ate a lista ficar ordenada

```python
# tabuleiro com 9 espacos
tabuleiro = [" "] * 9

# funcao que mostra o tabuleiro
def mostrar():
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])

# funcao que verifica se o jogador ganhou
def verificar(jogador):
    combinacoes = [[0,1,2],[3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]]
    for c in combinacoes:
        if tabuleiro[c[0]] == tabuleiro[c[1]] == tabuleiro[c[2]] == jogador:
            return True
    return False

# começa com o jogador x
jogador = "x"

# repete por ate 9 turnos
for turno in range(9):
    mostrar()
    posicao = int(input("jogador " + jogador + ", escolha uma posicao (0-8): "))
    tabuleiro[posicao] = jogador

    # verifica se ganhou
    if verificar(jogador):
        mostrar()
        print("jogador", jogador, "ganhou!")
        break

    # troca o jogador
    if jogador == "x":
        jogador = "o"
    else:
        jogador = "x"
else:
    print("empate!")
```

37. Funções Recursivas
o Implemente uma função recursiva para calcular o fatorial de um
número.

```python
# funcao que chama ela mesma para calcular o fatorial
def fatorial(n):
    # caso base: fatorial de 0 e 1
    if n == 0:
        return 1
    # chama a funcao de novo com n-1
    return n * fatorial(n - 1)

# pede um numero
numero = int(input("digite um numero: "))

# mostra o fatorial
print("fatorial:", fatorial(numero))
```

a funcao chama ela mesma com n menos 1 ate chegar em 0, ai multiplica tudo voltando

46. Calcular Média e Desvio Padrão
o Calcule a média e o desvio padrão de uma lista de números.


```python
# lista de numeros
numeros = [4, 8, 6, 5, 3]

# calcula a media
total = 0
for n in numeros:
    total = total + n
media = total / len(numeros)

# calcula o desvio padrao
soma_quadrados = 0
for n in numeros:
    soma_quadrados = soma_quadrados + (n - media) ** 2
desvio = (soma_quadrados / len(numeros)) ** 0.5

# mostra os resultados
print("media:", media)
print("desvio padrao:", desvio)
```

calcula a media somando tudo e dividindo pelo total, depois calcula o desvio como raiz da media dos quadrados das diferencas

47. Contar Números em Lista
o Conte quantas vezes cada número aparece em uma lista.


```python
# lista com numeros repetidos
numeros = [1, 2, 2, 3, 1, 4, 3, 3]

# dicionario para guardar a contagem
contagem = {}

# conta cada numero
for n in numeros:
    if n in contagem:
        contagem[n] = contagem[n] + 1
    else:
        contagem[n] = 1

# mostra o resultado
print(contagem)
```

48.Ordenação por Inserção
o Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
# lista para ordenar
numeros = [5, 3, 8, 1, 9, 2]

# percorre a lista do segundo elemento em diante
for i in range(1, len(numeros)):
    # guarda o elemento atual
    chave = numeros[i]
    j = i - 1

    # move os elementos maiores para frente
    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j = j - 1

    # coloca a chave no lugar certo
    numeros[j + 1] = chave

# mostra a lista ordenada
print("lista ordenada:", numeros)
```

pega cada elemento e o encaixa na posicao certa entre os anteriores ja ordenados

50. Cálculo de Frequência de Letras
o Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

```python
# pede um texto
texto = input("digite um texto: ")

# transforma o texto em caracteres
palavras = list(texto)
# cria um dicionario vazio
contagem = {}

# percorre cada palavra
for palavra in palavras:
    # se a palavra ja existe adiciona 1
    if palavra in contagem:
        contagem[palavra] = contagem[palavra] + 1
    # se nao existe coloca como 1
    else:
        contagem[palavra] = 1

# mostra o resultado
print(contagem)
```
